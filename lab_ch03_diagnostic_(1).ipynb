Part 1 hands you a merged dataset with five things wrong with it and asks you

````python
# -----------------------------------------------------------
# GUIDED — Run as-is
# Step 0a: groupby, map, apply
# -----------------------------------------------------------
import pandas as pd
import numpy as np

clean = pd.DataFrame({
    "country":   ["USA", "USA", "China", "China", "Brazil", "Brazil"],
    "year":      [2022, 2023, 2022, 2023, 2022, 2023],
    "gdp":       [25.7, 27.4, 17.9, 18.5, 1.92, 2.17],     # trillions USD
    "region":    ["Americas", "Americas", "Asia", "Asia", "Americas", "Americas"],
})

# .groupby(col) SPLITS the table into one group per distinct value, APPLIES an
# aggregation to each, and COMBINES the answers. Split -> apply -> combine.
print("Mean GDP by region:")
print(clean.groupby("region")["gdp"].mean().round(2))

# Note what came back: a Series whose INDEX is the region. groupby always turns
# the grouping column into the index. .reset_index() pushes it back to a column.
print("\nSeveral statistics at once — .agg():")
print(clean.groupby("country")["gdp"].agg(["count", "mean", "max"]).round(2))

# .map() applies a function to EVERY VALUE of a Series and returns a new one.
# `lambda x: ...` is a one-expression function with no name, handed to .map()
# so it can be run once per value.
clean["gdp_billions"] = clean["gdp"].map(lambda x: x * 1000)

# .map() also takes a DICT, which is how you recode labels.
clean["region_code"] = clean["region"].map({"Americas": "AM", "Asia": "AS"})

# .apply(axis=1) hands your function an entire ROW rather than one value.
clean["label"] = clean.apply(
    lambda row: f"{row['country']} {row['year']}: ${row['gdp']:.2f}T", axis=1
)
print("\n" + "\n".join(clean["label"]))
````

````python
# -----------------------------------------------------------
# GUIDED — Run as-is
# Step 0b: the loop helpers, and try / except
# -----------------------------------------------------------

countries = ["USA", "China", "Brazil"]
gdp       = [27.4, 18.5, 2.17]

# zip() walks two lists side by side, one pair per step.
for name, value in zip(countries, gdp):
    print(f"  {name:<8}${value:>6.2f}T")

# enumerate() walks a list while counting positions — `i` is 0, 1, 2 ...
# This is the form you will read in every multi-panel plotting cell.
print()
for i, name in enumerate(countries):
    print(f"  panel {i}: {name}")

# range(n) repeats n times. The throwaway `_` means "I do not need this value".
total = 0
for _ in range(3):
    total += 1
print(f"\nRepeated 3 times, total = {total}")

# try / except catches ONE SPECIFIC failure and carries on. Always name the
# exception you expect; a bare `except:` swallows real bugs and leaves you
# debugging blind.
raw = ["3.1", "2.5", "n/a", "4.5", ""]
cleaned = []
for item in raw:
    try:
        cleaned.append(float(item))
    except ValueError:
        cleaned.append(np.nan)      # np.nan is "missing number"
print(f"\nRaw:     {raw}")
print(f"Cleaned: {cleaned}")
````

````python
# -----------------------------------------------------------
# GUIDED — Run as-is
# Step 0c: BUILD IT — what a HEALTHY profile looks like
# -----------------------------------------------------------
# Part 1 hands you a corrupted version of a panel shaped exactly like this one.
# Run the same four checks on clean data first, so you know what "fine" reads
# like. These four are the whole EDA checklist:
#
#     STRUCTURE     — shape, dtypes, one row per what?
#     DISTRIBUTIONS — describe(): any impossible values?
#     COMPLETENESS  — nulls, and duplicate keys
#     RANGES        — does every column obey its domain constraint?

rng = np.random.default_rng(42)
n_c, yrs = 10, list(range(2000, 2024))
healthy = pd.DataFrame([
    {"country": c, "year": y,
     "gdp": rng.uniform(500, 25000),
     "population": rng.uniform(5, 1400),
     "life_expectancy": rng.uniform(55, 85),
     "trade_pct": rng.uniform(20, 85)}
    for c in [f"C{i}" for i in range(n_c)] for y in yrs
])

print("1. STRUCTURE")
print(f"   shape {healthy.shape} — expect {n_c} countries x {len(yrs)} years = {n_c * len(yrs)}")
print(f"   one row per country-year? duplicates: "
      f"{healthy.duplicated(subset=['country', 'year']).sum()}")

print("\n2. DISTRIBUTIONS")
print(healthy[["gdp", "population", "life_expectancy", "trade_pct"]]
      .describe().loc[["min", "50%", "max"]].round(1).to_string())

print("\n3. COMPLETENESS")
print(f"   nulls per column: {healthy.isna().sum().sum()} total")

print("\n4. RANGES — every column against its domain constraint")
constraints = {"gdp": (0, 1e6), "population": (0, 2000),
               "life_expectancy": (0, 120), "trade_pct": (0, 200)}
for col, (lo, hi) in constraints.items():
    bad = ((healthy[col] < lo) | (healthy[col] > hi)).sum()
    print(f"   {col:<18} outside [{lo}, {hi}]: {bad}")
````

````python
# -----------------------------------------------------------
# GIVEN — Run as-is. This generates the corrupted dataset.
# Do NOT modify this cell.
# -----------------------------------------------------------
np.random.seed(42)

countries = ["USA", "China", "India", "Brazil", "Germany",
             "Japan", "UK", "France", "Canada", "Australia"]
years = list(range(2000, 2024))
n_base = 240  # 10 countries x 24 years

# --- Source 1: Main dataset (clean) ---
rows = []
for c in countries:
    for y in years:
        gdp = np.random.uniform(500, 25000)  # GDP in millions USD
        pop = np.random.uniform(5, 1400)      # population in millions
        le = np.random.uniform(55, 85)         # life expectancy in years
        trade = np.random.uniform(20, 85)      # trade as % of GDP
        rows.append([c, y, gdp, pop, le, trade])

df = pd.DataFrame(rows, columns=["country", "year", "gdp", "population",
                                  "life_expectancy", "trade_pct"])

# --- corruption 1 ---
neg_idx = np.random.choice(df.index, size=10, replace=False)
df.loc[neg_idx, "gdp"] = df.loc[neg_idx, "gdp"] * -1

# --- corruption 2 ---
month_idx = np.random.choice(df.index, size=15, replace=False)
df.loc[month_idx, "life_expectancy"] = df.loc[month_idx, "life_expectancy"] * 12

# --- corruption 3 ---
dup_idx = np.random.choice(df.index, size=20, replace=False)
dup_rows = df.loc[dup_idx].copy()
# Slightly perturb duplicates so they aren't exact copies
dup_rows["gdp"] = dup_rows["gdp"] * np.random.uniform(0.98, 1.02, size=20)
df = pd.concat([df, dup_rows], ignore_index=True)

# --- corruption 4 ---
# Some rows stored as decimals (0.45) instead of percentages (45%)
decimal_idx = np.random.choice(df.index, size=30, replace=False)
df.loc[decimal_idx, "trade_pct"] = df.loc[decimal_idx, "trade_pct"] / 100
# Some rows have impossible values (data entry: 350%, 450%)
impossible_idx = np.random.choice(
    df.index.difference(decimal_idx), size=10, replace=False
)
df.loc[impossible_idx, "trade_pct"] = np.random.uniform(200, 500, size=10)

# --- corruption 5 ---
unit_idx = np.random.choice(df.index, size=50, replace=False)
df.loc[unit_idx, "gdp"] = df.loc[unit_idx, "gdp"] / 1000

# Shuffle so bugs aren't at the end
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print(f"Dataset shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
````

## Part 2: Distribution Shift Detection (EXTEND — 15 min)

A clean dataset today doesn't guarantee a clean dataset tomorrow. In production, the data generating process can change — a new data source, a policy shock, a pandemic. This section builds a diagnostic that detects when the distribution of incoming data has drifted from the training data.

````python
# -----------------------------------------------------------
# GIVEN — Split into training and inference sets
# Deliberately shift GDP in the inference set
# -----------------------------------------------------------

# Use cleaned data from Part 1
# Split proportionally, not at a fixed row number — the dataset is 260 rows
# before your FIX 3 removes duplicates, so any hard-coded index near 300
# leaves the inference set empty.
split = int(len(df_clean) * 0.6)
train = df_clean.iloc[:split].copy()
inference = df_clean.iloc[split:].copy()

# Simulate a distribution shift: inference GDP is 30% higher
# (e.g., new data source reports GDP in a different unit)
inference["gdp"] = inference["gdp"] * 1.3

print(f"Training set: {len(train)} rows")
print(f"Inference set: {len(inference)} rows")
print(f"\nTraining GDP mean: {train['gdp'].mean():.0f}")
print(f"Inference GDP mean: {inference['gdp'].mean():.0f}")
````

````python
# -----------------------------------------------------------
# GIVEN — run as-is: KDE overlays, training vs. inference
# Nothing to fill in. Read the four panels: which column's two curves
# have come apart, and which three sit on top of each other?
# -----------------------------------------------------------

numeric_cols = ["gdp", "population", "life_expectancy", "trade_pct"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    ax = axes[i]
    train[col].plot.kde(ax=ax, label="Training", color="#1976D2", linewidth=2)
    inference[col].plot.kde(ax=ax, label="Inference", color="#D32F2F",
                           linewidth=2, linestyle="--")
    ax.set_title(col, fontsize=13)
    ax.legend(frameon=False)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.suptitle("Distribution Shift Detection: Training vs. Inference",
             fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
````

````python
# -----------------------------------------------------------
# GIVEN — run as-is: Population Stability Index (PSI)
# Nothing to fill in. PSI puts a number on what the KDE panels show.
# Your work in this part is the interpretation question below it —
# read the four values first, and be ready to say which you trust.
# -----------------------------------------------------------

def compute_psi(expected, actual, n_bins=10):
    """Compute Population Stability Index between two distributions.

    PSI = sum( (actual_i - expected_i) * ln(actual_i / expected_i) )
    where actual_i and expected_i are the proportion of observations
    in bin i.

    Parameters
    ----------
    expected : array-like
        Reference distribution (training data).
    actual : array-like
        New distribution (inference data).
    n_bins : int
        Number of bins for the histogram.

    Returns
    -------
    float
        PSI value.
    """
    # Bin edges based on expected (training) distribution
    breakpoints = np.linspace(
        min(np.min(expected), np.min(actual)),
        max(np.max(expected), np.max(actual)),
        n_bins + 1
    )

    expected_counts = np.histogram(expected, bins=breakpoints)[0]
    actual_counts = np.histogram(actual, bins=breakpoints)[0]

    # Convert to proportions, add small epsilon to avoid division by zero
    eps = 1e-4
    expected_pct = expected_counts / len(expected) + eps
    actual_pct = actual_counts / len(actual) + eps

    psi = np.sum((actual_pct - expected_pct) * np.log(actual_pct / expected_pct))
    return psi


# Compute PSI for each numeric column
print("=== Population Stability Index (PSI) ===")
print(f"{'Column':<20} {'PSI':>8}  {'Interpretation'}")
print("-" * 55)

for col in numeric_cols:
    psi_val = compute_psi(train[col].values, inference[col].values)
    if psi_val < 0.1:
        interp = "Stable"
    elif psi_val < 0.25:
        interp = "Moderate shift"
    else:
        interp = "SIGNIFICANT SHIFT"
    print(f"{col:<20} {psi_val:>8.4f}  {interp}")
````

of the *same undrifted* data and use that distribution's upper tail as the
alarm level. The GDP column is the only one here whose PSI is large enough to
survive that comparison.

### Interpretation Questions

**Q1:** Why is detecting distribution shift important before deploying a model to production? What happens to predictions if the input distribution changes but the model doesn't know?

*Your answer here:* Detecting distribution shift is crucial before deploying a model because models learn patterns from the training data. If the input data distribution changes (drifts) in production, the model's learned patterns may no longer be accurate or relevant. This can lead to a significant degradation in prediction performance, making the model's outputs unreliable or even harmful without the model 'knowing' it's making poor predictions. For example, a credit risk model trained on data before a major economic recession might perform poorly if deployed during or after one, as consumer spending and default patterns would have shifted, violating the model's assumptions.

**Q2:** A model trained on pre-COVID economic data is deployed in 2021. Which columns in our dataset likely shifted most? How would you design a monitoring system to catch this?

*Your answer here:* A model trained on pre-COVID economic data deployed in 2021 would likely see significant shifts in `gdp` (due to economic contractions and subsequent recovery, policy interventions), `trade_pct` (disruptions to global supply chains, changes in trade policies), and potentially `population` (migration patterns, mortality rates from the pandemic). `life_expectancy` might also show a temporary dip due to the direct health impact of the pandemic. To catch this, a monitoring system would continuously calculate PSI or other drift metrics for key features (`gdp`, `trade_pct`, `population`, `life_expectancy`) between recent production data and the training baseline. Alarms would be triggered if PSI crosses a predefined, statistically validated threshold (one not based on arbitrary cutoffs like 0.1/0.25 but derived from a null distribution for the given sample sizes), and an alert system would notify data scientists for investigation and potential model retraining.

**Q2:** In a production pipeline processing 1M rows daily, would you rely on manual or automated EDA? What's your monitoring strategy?

For a production pipeline processing 1 million rows daily, it would be impossible and inefficient to rely on manual EDA. An **automated EDA and monitoring strategy** is essential. My strategy would involve:

1.  **Automated Data Quality Checks:** Implement pre-defined rules for each feature (e.g., `check_impossible_values` for domain constraints, `not null` checks) that run on every new batch of data. Violations would trigger immediate alerts.
2.  **Distribution Shift Detection:** Employ metrics like Population Stability Index (PSI) or Jensen-Shannon Divergence (`detect_distribution_shift`) for all critical numerical and categorical features. These would compare the distribution of incoming production data against a baseline (e.g., training data or a recent stable period). Thresholds would be set for PSI values to distinguish stable, moderate, and significant shifts.

````python
%%writefile eda_utils.py
"""EDA utilities for ECON 5200 labs.

Functions for data quality checking, distribution shift detection,
and automated EDA summaries.
"""

import numpy as np
import pandas as pd


def check_impossible_values(df, constraints):
    """Check columns against domain constraints and flag violations.

    Parameters
    ----------
    df : pd.DataFrame
        Input data to validate.
    constraints : dict
        Column name -> (min_val, max_val) or callable.
        If a tuple, flags rows outside [min_val, max_val].
        If a callable, flags rows where callable(value) returns False.

    Returns
    -------
    pd.DataFrame
        DataFrame with boolean columns '{col}_flagged' for each
        constrained column. True = violation detected.

    Examples
    --------
    >>> constraints = {
    ...     'gdp': (0, 100000),
    ...     'life_expectancy': (0, 100),
    ...     'trade_pct': lambda x: (x >= 0) & (x <= 100),
    ... }
    >>> flagged = check_impossible_values(df, constraints)
    """
    result = df.copy()

    for col, rule in constraints.items():
        if col not in df.columns:
            continue

        if callable(rule):
            result[f"{col}_flagged"] = ~rule(df[col])
        else:
            min_val, max_val = rule
            result[f"{col}_flagged"] = (
                (df[col] < min_val) | (df[col] > max_val)
            )

    flag_cols = [c for c in result.columns if c.endswith("_flagged")]
    n_flagged = result[flag_cols].any(axis=1).sum()
    print(f"Flagged {n_flagged} rows with constraint violations.")
    for fc in flag_cols:
        n = result[fc].sum()
        if n > 0:
            print(f"  {fc}: {n} violations")

    return result


def detect_distribution_shift(train_col, inference_col, n_bins=10):
    """Compute Population Stability Index between two distributions.

    PSI measures how much a distribution has shifted from a reference
    (training) distribution. Used in ML pipelines to detect data drift.

    PSI = sum( (actual_i - expected_i) * ln(actual_i / expected_i) )

    Parameters
    ----------
    train_col : array-like
        Reference distribution (training data).
    inference_col : array-like
        New distribution (inference/production data).
    n_bins : int, default 10
        Number of bins for histogram comparison.

    Returns
    -------
    float
        PSI value.
    str
        Interpretation: 'stable' (< 0.1), 'moderate_shift' (0.1-0.25),
        or 'significant_shift' (> 0.25).

    Examples
    --------
    >>> psi, interp = detect_distribution_shift(train['gdp'], test['gdp'])
    >>> print(f'PSI = {psi:.4f} ({interp})')
    """
    train_arr = np.asarray(train_col, dtype=float)
    inf_arr = np.asarray(inference_col, dtype=float)

    breakpoints = np.linspace(
        min(train_arr.min(), inf_arr.min()),
        max(train_arr.max(), inf_arr.max()),
        n_bins + 1,
    )

    expected_counts = np.histogram(train_arr, bins=breakpoints)[0]
    actual_counts = np.histogram(inf_arr, bins=breakpoints)[0]

    eps = 1e-4
    expected_pct = expected_counts / len(train_arr) + eps
    actual_pct = actual_counts / len(inf_arr) + eps

    psi = float(np.sum(
        (actual_pct - expected_pct) * np.log(actual_pct / expected_pct)
    ))

    if psi < 0.1:
        interpretation = "stable"
    elif psi < 0.25:
        interpretation = "moderate_shift"
    else:
        interpretation = "significant_shift"

    return psi, interpretation


def eda_summary(df):
    """Run a full EDA checklist and return a structured report.

    Checks for: missing values, duplicates, numeric outliers (IQR method),
    negative values in typically-positive columns, and basic shape info.

    Parameters
    ----------
    df : pd.DataFrame
        Input data to profile.

    Returns
    -------
    dict
        Structured report with keys: 'shape', 'dtypes', 'missing',
        'duplicates', 'numeric_summary', 'outlier_counts'.

    Examples
    --------
    >>> report = eda_summary(df)
    >>> print(report['outlier_counts'])
    """
    report = {}

    # Basic info
    report["shape"] = df.shape
    report["dtypes"] = df.dtypes.to_dict()

    # Missing values
    missing = df.isnull().sum()
    report["missing"] = missing[missing > 0].to_dict()

    # Exact duplicates
    report["duplicates"] = int(df.duplicated().sum())

    # Numeric summary
    numeric_df = df.select_dtypes(include=[np.number])
    report["numeric_summary"] = numeric_df.describe().to_dict()

    # Outlier detection (IQR method)
    outlier_counts = {}
    for col in numeric_df.columns:
        q1 = numeric_df[col].quantile(0.25)
        q3 = numeric_df[col].quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        n_outliers = int(
            ((numeric_df[col] < lower) | (numeric_df[col] > upper)).sum()
        )
        if n_outliers > 0:
            outlier_counts[col] = n_outliers
    report["outlier_counts"] = outlier_counts

    # Print summary
    print(f"Shape: {report['shape']}")
    print(f"Duplicates: {report['duplicates']}")
    if report["missing"]:
        print(f"Missing values: {report['missing']}")
    else:
        print("Missing values: None")
    if report["outlier_counts"]:
        print(f"Outliers (IQR): {report['outlier_counts']}")
    else:
        print("Outliers (IQR): None detected")

    return report
````

````python
# -----------------------------------------------------------
# 🤖 AI EXPANSION — Co-Pilot required
# Copy the P.R.I.M.E. prompt above into Claude, then paste
# the generated code here. Run it and verify.
# -----------------------------------------------------------

# [Prep] Act as an expert Python data scientist specializing
# in data-pipeline monitoring and distribution shift.
#
# [Request] I just completed a diagnosis-first lab where I found
# and fixed five planted corruptions in a country panel using
# EDA alone (negative GDP, life expectancy stored in months, duplicates,
# mixed percentage units, a GDP unit mismatch)
# * Measured distribution shift with PSI: GDP PSI = [YOUR VALUE] between
#   training and inference
# * Compared manual EDA with ydata-profiling and named what each missed
# * Wrote eda_utils.py with impossible-value, PSI and summary functions
# * Built an interactive pipeline-health dashboard
#
# The data is already in this runtime, uploaded next to
# eda_utils.py (the save cell above writes and zips it):
#   pipeline_health_dirty.csv   260 rows, all five bugs intact
#   pipeline_health_clean.csv   230 rows, after the fixes
#   training.csv                138 rows, reference distribution
#   inference.csv                92 rows, gdp already x1.3
# Load them with pandas. Do NOT assume df / df_clean / train /
# inference already exist -- I may be running this in a fresh
# notebook with only those files.
#
# [Iterate] Use ipywidgets, plotly and pandas. `import eda_utils`
# from the same folder and call it as it is -- do not reimplement
# PSI. Note the shape: detect_distribution_shift(train_col,
# inference_col, n_bins=10) takes two COLUMNS, not two DataFrames,
# and returns a (psi, interpretation) pair. Passing frames raises
# "could not convert string to float". No Streamlit, no new .py
# file, no ydata-profiling.
#
# [Mechanism Check] Add inline comments explaining:
#   - how the PSI bins are built: read eda_utils first and say
#     which sample defines the bin edges -- do not assume it is
#     the training set
#   - why the bin proportions get a small epsilon before the log
#   - why 0.10 and 0.25 are conventions, not statistical tests
#   - which of the five corruptions a PSI check would miss
#
# [Evaluate] Explain what the dashboard catches that automated
# profiling did not, and what it still cannot see.

# PASTE AI-GENERATED CODE BELOW:
# ============================================================
# PIPELINE-HEALTH DASHBOARD  (ipywidgets + plotly + pandas)
# Fresh-runtime safe: loads everything from disk, assumes no
# df / df_clean / train / inference in memory.
# ============================================================
import os, sys, re, glob, zipfile, inspect, importlib
import numpy as np
import pandas as pd
import ipywidgets as widgets
import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import display

try:                                   # Colab needs its own plotly renderer
    import google.colab  # noqa: F401
    pio.renderers.default = "colab"
except ImportError:
    pass

# ---------- 0. Locate + load files ----------------------------------------
FILES = {"dirty": "pipeline_health_dirty.csv", "clean": "pipeline_health_clean.csv",
         "train": "training.csv", "infer": "inference.csv"}
# Corrected EXPECTED_ROWS to match actual dataframe sizes after fixes and splits
EXPECTED_ROWS = {"dirty": 260, "clean": 240, "train": 144, "infer": 96}
NEEDED = list(FILES.values()) + ["eda_utils.py"]
SEARCH_DIRS = [os.getcwd(), "/content"]

def _has_all(d):
    return all(os.path.exists(os.path.join(d, f)) for f in NEEDED)

def find_data_dir():
    for d in SEARCH_DIRS:
        if _has_all(d):
            return d
    # The save cell zips the files; if only the zip is present, extract it.
    for d in SEARCH_DIRS:
        for z in glob.glob(os.path.join(d, "*.zip")):
            with zipfile.ZipFile(z) as zf:
                zf.extractall(d)
            for cand in [d] + glob.glob(os.path.join(d, "*/")):
                if _has_all(cand):
                    return cand
    raise FileNotFoundError(f"Need {NEEDED} in one of {SEARCH_DIRS} (or a zip there).")

DATA_DIR = os.path.abspath(find_data_dir())
if DATA_DIR not in sys.path:
    sys.path.insert(0, DATA_DIR)

import eda_utils
importlib.reload(eda_utils)            # pick up edits to eda_utils.py without restarting

frames = {k: pd.read_csv(os.path.join(DATA_DIR, f)) for k, f in FILES.items()}
dirty, clean, train, infer = (frames[k] for k in ("dirty", "clean", "train", "infer"))
for k, df in frames.items():
    flag = "" if len(df) == EXPECTED_ROWS[k] else f"   <-- expected {EXPECTED_ROWS[k]}"
    print(f"{FILES[k]:<28} {df.shape}{flag}")

# PSI runs only on columns numeric in BOTH splits.
num_cols = [c for c in train.columns
            if c in infer.columns
            and pd.api.types.is_numeric_dtype(train[c])
            and pd.api.types.is_numeric_dtype(infer[c])]
if not num_cols:
    raise ValueError("No shared numeric columns between training.csv and inference.csv.")
PROBE_COL = "gdp" if "gdp" in num_cols else num_cols[0]

# ---------- 1. Read eda_utils before trusting it --------------------------
# MECHANISM CHECK — which sample defines the bin edges?
# This cell does NOT assume the answer. It prints the lines of
# detect_distribution_shift that build bins/edges/epsilon, then runs an
# argument-swap probe. How to read the result:
#   * PSI = sum((q - p) * ln(q / p)) is symmetric in p and q for FIXED
#     edges. So if psi(train, infer) != psi(infer, train), the edges (or
#     the epsilon step) depend on argument order -> they come from one
#     argument, normally the first (reference). Confirm in the printed lines,
#     e.g. np.quantile(train_col, ...) or np.histogram_bin_edges(train_col).
#   * If the two are equal, edges are built from something symmetric:
#     pooled data (np.concatenate([train_col, inference_col])) or fixed values.
# Why it matters here: inference gdp is x1.3. If edges come from training
# min/max without -inf/+inf end bins, inference values above the training
# max fall outside every bin; np.histogram silently drops them, so the
# shift partly disappears from the PSI instead of inflating it. Quantile
# edges from training put ~1/n_bins of training in each bin by construction,
# so all of the sampling noise then sits in the inference proportions.
_src = inspect.getsource(eda_utils.detect_distribution_shift)
_pat = re.compile(r"quantile|percentile|histogram|linspace|qcut|\bcut\(|edge|bins?\b|eps|1e-|clip|inf", re.I)
print("\n--- eda_utils.detect_distribution_shift: binning / epsilon lines ---")
for i, line in enumerate(_src.splitlines(), 1):
    if _pat.search(line):
        print(f"{i:>4}: {line}")
print("(full source: print(inspect.getsource(eda_utils.detect_distribution_shift)))")

def psi_of(a, b, n_bins):
    """Thin wrapper: eda_utils takes two COLUMNS (Series), returns (psi, interpretation).
    NaNs are dropped before the call; remove .dropna() at the call sites if
    eda_utils is supposed to handle missing values itself."""
    psi, interp = eda_utils.detect_distribution_shift(a, b, n_bins=n_bins)
    return float(psi), interp

_fwd = psi_of(train[PROBE_COL].dropna(), infer[PROBE_COL].dropna(), 10)[0]
_rev = psi_of(infer[PROBE_COL].dropna(), train[PROBE_COL].dropna(), 10)[0]
print(f"\nSwap probe on '{PROBE_COL}': psi(train, infer)={_fwd:.4f}  psi(infer, train)={_rev:.4f}")
print("-> symmetric: edges pooled or fixed, not from one sample alone" if not np.isclose(_fwd, _rev, rtol=1e-9, atol=1e-12)
      else "-> symmetric: edges pooled or fixed, not from one sample alone")

# MECHANISM CHECK — why an epsilon before the log?
# A bin with zero share in either sample gives ln(0) or division by zero,
# so one empty bin makes PSI inf/nan. Adding (or clipping to) a small eps
# keeps it finite. It is a smoothing choice, not neutral: an empty bin
# contributes about (q - eps) * ln(q / eps), which grows as eps shrinks.
# With 92 inference rows and 10 bins (~9 rows/bin), empty or near-empty
# bins are realistic, so check what eps eda_utils uses in the lines above.
# If you see none, zero bins will surface below as inf/nan.

# MECHANISM CHECK — why 0.10 / 0.25 are conventions, not tests
# They are rules of thumb from credit-scorecard monitoring. They carry no
# sampling distribution, and they ignore sample size and n_bins even
# though PSI's noise floor depends on both. Under no shift, PSI is roughly
# (n_bins - 1) * (1/n_train + 1/n_infer). Here that is
# 9 * (1/138 + 1/92) = 0.163, already above 0.10. A quick simulation
# (lognormal data, no shift, 10 quantile bins, 5,000 draws, my own PSI
# implementation, not yours) gave mean PSI about 0.17, with about 80% of
# draws above 0.10 and about 16% above 0.25. So the permutation baseline
# below uses YOUR function to measure what noise alone produces on these
# exact sample sizes. It shuffles train+inference rows, which assumes rows
# are exchangeable; that is shaky if the split is by time or country.

# MECHANISM CHECK — which of the five corruptions would PSI miss?
# Structurally, as wired here: all of them. PSI compares training vs
# inference, and 138 + 92 = 230 = the clean frame's row count, which
# suggests both splits come from the CLEANED data. The dirty-frame bugs
# never enter this comparison; the only thing PSI can see is the injected
# gdp x1.3. Even if you pointed PSI at dirty vs clean, it would be blind to:
# duplicate rows (same shape, only more mass), a corruption present in both
# samples in the same proportion, string/non-numeric entries in numeric
# columns (dropped or unparseable, never binned), missing values (dropped
# before binning), wrong-but-plausible values that stay inside the same
# bins, and anything in a categorical/key column (e.g. inconsistent country
# names), since PSI here is numeric-only. A handful of impossible values
# (e.g. a negative population) in a 260-row frame moves bin shares by only
# ~1-2%, well under the noise floor above. Map your five bugs onto this list.

# ---------- 2. PSI table + permutation baseline (uses eda_utils only) -----
_psi_cache = {}

def psi_table(n_bins, n_perm, seed=0):
    key = (n_bins, n_perm)
    if key in _psi_cache:
        return _psi_cache[key]
    rng = np.random.default_rng(seed)
    rows = []
    for c in num_cols:
        a = train[c].dropna().reset_index(drop=True)
        b = infer[c].dropna().reset_index(drop=True)
        row = dict(column=c, n_train=len(a), n_infer=len(b))
        try:
            row["psi"], row["interpretation"] = psi_of(a, b, n_bins)
        except Exception as e:
            row.update(psi=np.nan, interpretation=f"ERROR: {e}")
            rows.append(row)
            continue
        null = []
        pooled = np.concatenate([a.to_numpy(), b.to_numpy()])
        for _ in range(n_perm):
            p = rng.permutation(pooled)
            try:
                null.append(psi_of(pd.Series(p[:len(a)]), pd.Series(p[len(a):]), n_bins)[0])
            except Exception:
                pass
        null = np.array([v for v in null if np.isfinite(v)])
        if len(null):
            row["null_mean"] = null.mean()
            row["null_p95"] = np.quantile(null, 0.95)
            row["perm_p"] = (1 + (null >= row["psi"]).sum()) / (1 + len(null))
        rows.append(row)
    t = (pd.DataFrame(rows)
           .reindex(columns=["column", "psi", "interpretation", "null_mean",
                             "null_p95", "perm_p", "n_train", "n_infer"])
           .sort_values("psi", ascending=False, na_position="last")
           .reset_index(drop=True))
    _psi_cache[key] = t
    return t

# ---------- 3. Panel 1: train vs inference overlay -------------------------
col_dd = widgets.Dropdown(options=num_cols, value=PROBE_COL, description="Column")
bins_sl = widgets.IntSlider(value=10, min=3, max=30, description="n_bins")
view_tb = widgets.ToggleButtons(options=["Histogram", "ECDF"], value="Histogram")
out1 = widgets.Output()

def _ecdf(s):
    x = np.sort(s.to_numpy())
    return x, np.arange(1, len(x) + 1) / len(x)

def draw_overlay(*_):
    c, nb = col_dd.value, bins_sl.value
    a, b = train[c].dropna(), infer[c].dropna()
    try:
        psi, interp = psi_of(a, b, nb)
        title = f"{c}: PSI = {psi:.3f} ({interp}), n_bins = {nb}"
    except Exception as e:
        title = f"{c}: PSI failed: {e}"
    fig = go.Figure()
    if view_tb.value == "Histogram":
        # Display bins only: equal-width over the pooled range. These are NOT
        # the edges eda_utils uses for PSI (see Section 1).
        lo, hi = float(min(a.min(), b.min())), float(max(a.max(), b.max()))
        xb = dict(start=lo, end=hi, size=(hi - lo) / nb if hi > lo else 1.0)
        fig.add_trace(go.Histogram(x=a, name=f"train (n={len(a)})", histnorm="probability", opacity=0.55, xbins=xb))
        fig.add_trace(go.Histogram(x=b, name=f"inference (n={len(b)})", histnorm="probability", opacity=0.55, xbins=xb))
        fig.update_layout(barmode="overlay", yaxis_title="share of rows")
    else:
        # ECDF: bin-free, so a pure scale shift (x1.3) shows up as a horizontal stretch.
        for s, name in ((a, "train"), (b, "inference")):
            x, y = _ecdf(s)
            fig.add_trace(go.Scatter(x=x, y=y, mode="lines", line_shape="hv", name=f"{name} (n={len(s)})"))
        fig.update_layout(yaxis_title="cumulative share")
    fig.update_layout(title=title, xaxis_title=c, height=420)
    with out1:
        out1.clear_output(wait=True)
        fig.show()

for w in (col_dd, bins_sl, view_tb):
    w.observe(draw_overlay, names="value")
tab1 = widgets.VBox([widgets.HBox([col_dd, bins_sl]), view_tb, out1])

# ---------- 4. Panel 2: PSI threshold -> flagged columns -------------------
thr_sl = widgets.FloatSlider(value=0.10, min=0.0, max=0.5, step=0.01,
                             description="PSI >=", readout_format=".2f")
nperm_sl = widgets.IntSlider(value=200, min=0, max=1000, step=50, description="permutations")
out2 = widgets.Output()

def draw_flags(*_):
    thr = thr_sl.value
    with out2:
        out2.clear_output(wait=True)
        print(f"Computing (n_bins={bins_sl.value}, permutations={nperm_sl.value})...")
        t = psi_table(bins_sl.value, nperm_sl.value)
        out2.clear_output(wait=True)
        flagged = t[t["psi"] >= thr]
        print(f"n_bins={bins_sl.value} (shared with panel 1)   threshold={thr:.2f}   "
              f"flagged {len(flagged)}/{len(t)}: {list(flagged['column'])}")
        if nperm_sl.value:
            print("null_p95 = 95th pct of PSI under shuffled rows (noise floor for these n); "
                  "perm_p = share of shuffles with PSI >= observed.")
        display(t.style
                 .apply(lambda r: ["background-color:#fde2e1" if r["psi"] >= thr else "" for _ in r], axis=1)
                 .format({"psi": "{:.3f}", "null_mean": "{:.3f}", "null_p95": "{:.3f}", "perm_p": "{:.3f}"},
                         na_rep="-"))

thr_sl.observe(draw_flags, names="value")          # cached: instant
nperm_sl.observe(draw_flags, names="value")        # recomputes once per new value
bins_sl.observe(draw_flags, names="value")
tab2 = widgets.VBox([widgets.HBox([thr_sl, nperm_sl]), out2])

# ---------- 5. Panel 3: editable constraints -> check_impossible_values ----
con_cols = [c for c in clean.columns if pd.api.types.is_numeric_dtype(clean[c])]
con_rows = {}
for c in con_cols:
    con_rows[c] = (widgets.Checkbox(value=False, description=c, indent=False, layout=widgets.Layout(width="200px")),
                   widgets.Text(placeholder="min (blank = none)", layout=widgets.Layout(width="160px")),
                   widgets.Text(placeholder="max (blank = none)", layout=widgets.Layout(width="160px")))
frame_tb = widgets.ToggleButtons(options=["dirty", "clean", "both"], value="both", description="Frame")
run3 = widgets.Button(description="Run check", button_style="primary")
prefill3 = widgets.Button(description="Prefill from clean range")
out3 = widgets.Output()

print("\n--- eda_utils.check_impossible_values signature / docstring ---")
print(inspect.signature(eda_utils.check_impossible_values))
print(inspect.getdoc(eda_utils.check_impossible_values) or "(no docstring)")

def _num(txt, default):
    txt = txt.strip()
    return default if txt == "" else float(txt)

def build_constraints():
    cons = {}
    for c, (en, mn, mx) in con_rows.items():
        if en.value:
            cons[c] = (_num(mn.value, -np.inf), _num(mx.value, np.inf))
    return cons

def call_check(df, cons):
    # Tries {col: (min, max)} first, then {col: {"min":..,"max":..}}. If your
    # eda_utils uses another shape, the docstring printed above says which;
    # change this one function.
    try:
        return eda_utils.check_impossible_values(df, cons)
    except Exception as e1:
        try:
            return eda_utils.check_impossible_values(df, {c: {"min": lo, "max": hi} for c, (lo, hi) in cons.items()})
        except Exception as e2:
            raise RuntimeError(f"tuple form: {e1!r}\ndict form:  {e2!r}")

def run_check(*_):
    with out3:
        out3.clear_output(wait=True)
        try:
            cons = build_constraints()
        except ValueError as e:
            print(f"Bad number in a min/max box: {e}")
            return
        if not cons:
            print("Tick at least one column and set a min and/or max.")
            return
        print("constraints:", cons)
        # Dirty frame is passed AS-IS: if a corrupted column is object dtype,
        # the error below is itself a finding (non-numeric values present).
        for name in (["dirty", "clean"] if frame_tb.value == "both" else [frame_tb.value]):
            print(f"\n=== {name} ({len(frames[name])} rows) ===")
            try:
                res = call_check(frames[name], cons)
                display(res)
            except Exception as e:
                print(f"check_impossible_values failed on {name}:\n{e}")

def prefill(*_):
    # CIRCULAR: the clean frame's own range flags nothing on the clean frame
    # by construction. Use it as a dirty-vs-clean diff, then replace with
    # domain bounds (population >= 0, rates in [0, 100], ...).
    for c, (en, mn, mx) in con_rows.items():
        en.value, mn.value, mx.value = True, repr(float(clean[c].min())), repr(float(clean[c].max()))

run3.on_click(run_check)
prefill3.on_click(prefill)
tab3 = widgets.VBox([widgets.HBox([frame_tb, run3, prefill3])]
                    + [widgets.HBox(list(w)) for w in con_rows.values()] + [out3])

# ---------- 6. Panel 4: before / after (dirty vs clean) --------------------
def column_diff():
    rows = []
    for c in list(dirty.columns) + [c for c in clean.columns if c not in dirty.columns]:
        r = {"column": c}
        for name, df in (("dirty", dirty), ("clean", clean)):
            if c not in df.columns:
                r[f"{name}_dtype"] = "(missing)"
                continue
            s = df[c]
            coerced = pd.to_numeric(s, errors="coerce")
            r[f"{name}_dtype"] = str(s.dtype)
            r[f"{name}_nulls"] = int(s.isna().sum())
            # non-null values that are not parseable as numbers
            r[f"{name}_non_numeric"] = int((s.notna() & coerced.isna()).sum())
            r[f"{name}_min"] = coerced.min()
            r[f"{name}_max"] = coerced.max()
            r[f"{name}_nunique"] = int(s.nunique())
        rows.append(r)
    return pd.DataFrame(rows)

_ba_opts = [c for c in dirty.columns if c in clean.columns]
ba_dd = widgets.Dropdown(options=_ba_opts, value=_ba_opts[0], description="Column")
out4_top, out4_plot = widgets.Output(), widgets.Output()
out4_sum_d, out4_sum_c = widgets.Output(), widgets.Output()

def draw_before_after_static():
    with out4_top:
        out4_top.clear_output()
        print(f"rows: dirty {len(dirty)} -> clean {len(clean)}  (removed {len(dirty) - len(clean)})")
        print(f"exact duplicate rows: dirty {int(dirty.duplicated().sum())}, clean {int(clean.duplicated().sum())}")
        display(column_diff())
    for out, name in ((out4_sum_d, "dirty"), (out4_sum_c, "clean")):
        with out:
            out.clear_output()
            print(f"eda_summary({name})")
            try:
                res = eda_utils.eda_summary(frames[name])
                if res is not None:
                    display(res)
            except Exception as e:
                print(f"failed: {e!r}")

def draw_ba_plot(*_):
    c = ba_dd.value
    d, k = pd.to_numeric(dirty[c], errors="coerce"), pd.to_numeric(clean[c], errors="coerce")
    fig = go.Figure()
    if d.notna().any() or k.notna().any():
        fig.add_trace(go.Histogram(x=d.dropna(), name=f"dirty (numeric n={d.notna().sum()})", histnorm="probability", opacity=0.55))
        fig.add_trace(go.Histogram(x=k.dropna(), name=f"clean (numeric n={k.notna().sum()})", histnorm="probability", opacity=0.55))
        fig.update_layout(barmode="overlay", yaxis_title="share of rows")
    else:  # categorical column: compare value counts
        vc = pd.concat([dirty[c].value_counts().rename("dirty"), clean[c].value_counts().rename("clean")], axis=1).fillna(0)
        vc = vc.sort_values("dirty", ascending=False).head(30)
        fig.add_trace(go.Bar(x=vc.index.astype(str), y=vc["dirty"], name="dirty"))
        fig.add_trace(go.Bar(x=vc.index.astype(str), y=vc["clean"], name="clean"))
        fig.update_layout(barmode="group", yaxis_title="count (top 30 by dirty)")
    fig.update_layout(title=f"{c}: dirty vs clean", height=400)
    with out4_plot:
        out4_plot.clear_output(wait=True)
        fig.show()

ba_dd.observe(draw_ba_plot, names="value")
tab4 = widgets.VBox([out4_top, ba_dd, out4_plot, widgets.HBox([out4_sum_d, out4_sum_c])])

# ---------- 7. Assemble ----------------------------------------------------
tabs = widgets.Tab(children=[tab1, tab2, tab3, tab4])
for i, title in enumerate(["Train vs inference", "PSI flags", "Constraints", "Before / after"]):
    tabs.set_title(i, title)
display(tabs)

draw_overlay()
draw_flags()
draw_before_after_static()
draw_ba_plot()
````
